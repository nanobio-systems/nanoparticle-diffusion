```python
from pathlib import Path
import sys

def find_project_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    for path in [start, *start.parents]:
        if (path / "src").is_dir() and (path / "notebooks").is_dir():
            return path
    raise RuntimeError("Could not find project root containing src/ and notebooks/")

PROJECT_ROOT = find_project_root()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(f"Project root: {PROJECT_ROOT}")
```

```python
import numpy as np

from src.physics import (
    diffusion_coefficient,
    brownian_step_size,
)

from src.simulation import (
    simulate_particle,
)

from src.analysis import (
    calculate_msd,
)

from src.constants import (
    DEFAULT_DT,
    DEFAULT_RADIUS,
    DEFAULT_TEMPERATURE,
    DEFAULT_VISCOSITY,
)
```

```python
D = diffusion_coefficient(
    DEFAULT_TEMPERATURE,
    DEFAULT_VISCOSITY,
    DEFAULT_RADIUS,
)

step = brownian_step_size(
    D,
    DEFAULT_DT,
)

print(D)

print(step)
```

```python
x, y = simulate_particle(
    1000,
    step,
)

print(x.shape)

print(y.shape)
```

```python
lag_times, msd = calculate_msd(
    x.reshape(1, -1),
    y.reshape(1, -1),
)

print("Lag times shape:", lag_times.shape)
print("MSD shape:", msd.shape)
print("First MSD value:", msd[0])
```

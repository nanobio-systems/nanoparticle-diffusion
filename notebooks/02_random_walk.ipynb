```python
from pathlib import Path
import sys

def find_project_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    for path in [start, *start.parents]:
        if (path / "src").is_dir() and (path / "notebooks").is_dir():
            return path
    raise RuntimeError("Could not find project root containing src/ and notebooks/")

PROJECT_ROOT = find_project_root()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(f"Project root: {PROJECT_ROOT}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
steps = 1000

x = np.zeros(steps)

y = np.zeros(steps)
```

```python
for i in range(1, steps):

    dx = np.random.normal()

    dy = np.random.normal()

    x[i] = x[i-1] + dx

    y[i] = y[i-1] + dy
```

```python
plt.figure(figsize=(6,6))

plt.plot(x,y)

plt.scatter(x[0],y[0],label="Start")

plt.scatter(x[-1],y[-1],label="End")

plt.legend()

plt.xlabel("x")

plt.ylabel("y")

plt.title("Single Nanoparticle Random Walk")

plt.axis("equal")

plt.show()
```

```python
distance = np.sqrt(x**2 + y**2)

plt.figure(figsize=(8,4))

plt.plot(distance)

plt.xlabel("Step")

plt.ylabel("Distance")

plt.title("Distance from Starting Point")

plt.show()
```

```python
print("Number of steps:", steps)

print("Final x position:", x[-1])

print("Final y position:", y[-1])

print("Maximum distance:", np.max(distance))
```
